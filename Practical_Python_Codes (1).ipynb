```python
import sys

def main():
  
  total_args = len(sys.argv)
  
  print(f"Program Name: {sys.argv[0]}")
 
  print(f"Total Number of Arguments: {total_args}")

  print("Arguments Passed:", sys.argv)

  
  total_sum = 0
  numeric_count = 0

  
  for arg in sys.argv[1:]:
    try:
      
      total_sum += float(arg)
      numeric_count += 1
    except ValueError:
      
      pass

  if numeric_count > 0:
    print(f"Sum of numeric arguments: {total_sum}")
  else:
    print("No numeric arguments provided to calculate a sum.")


if __name__ == "__main__":
  main()
```

```python
import math

def is_prime(num):
  
  if num <= 1:
    return False

  
  limit = int(math.sqrt(num)) + 1
  for i in range(2, limit):
    # If num is divisible by i, return False
    if num % i == 0:
      return False

  return True

def main():
  print("--- Task 1: Check if a Single Number is Prime ---")
  number = int(input("Enter a number to check: "))

  if is_prime(number):
    print(f"{number} is a prime number[cite: 6, 7].")
  else:
    print(f"{number} is not a prime number[cite: 6, 7].")

  print("\n--- Task 2: Find Prime Numbers within a Range ---")
  start = int(input("Enter the starting value of the range: "))
  end = int(input("Enter the ending value of the range: "))

  print(f"\nPrime numbers between {start} and {end}:")
  prime_count = 0

  
  for num in range(start, end + 1):
    
    if is_prime(num):
      print(num, end=" ")
      prime_count += 1

  print(
      f"\n\nTotal prime numbers found in the given range: {prime_count}"
  )


if __name__ == "__main__":
  main()
```

```python
import math
import random

def basic_math_operations():
  num1 = 16
  num2 = 4
  print(f"Square root of {num1}: {math.sqrt(num1)}")
  print(f"{num1} raised to the power of {num2}: {math.pow(num1, num2)}")
  print(f"Sine of {num1} degrees: {math.sin(math.radians(num1))}")
  print(f"Cosine of {num2} degrees: {math.cos(math.radians(num2))}")
  print(f"Factorial of {num2}: {math.factorial(num2)}")

def random_number_operations():
  print("\nRandom float between 0 and 1:", random.random())
  print("Random integer between 10 and 50:", random.randint(10, 50))
  print("Random choice from a list:", random.choice([2, 4, 6, 8, 10]))

  random_list = [1, 2, 3, 4, 5]
  random.shuffle(random_list)
  print("Shuffled list:", random_list)
  print("Random sample of 3 elements from a list:", random.sample(random_list, 3))

if __name__ == "__main__":
  print("Basic Math Operations:")
  basic_math_operations()
  print("\nRandom Number Operations:")
  random_number_operations()
```

```python

numbers = [1, 2, 3, 4, 5]  

squares = list(map(lambda x: x**2, numbers))  
evens = list(filter(lambda x: x % 2 == 0, numbers)) 

print("Original list:", numbers)
print("Squares using lambda:", squares)  
print("Even numbers using lambda:", evens)  


squares_lc = [x**2 for x in numbers]  
evens_lc = [x for x in numbers if x % 2 == 0]  

print("Squares using list comprehension:", squares_lc)  
print("Evens using list comprehension:", evens_lc)  
```

```python

def add(a, b):
  return a + b

def greet(first_name, last_name):
  return f"Hello, {first_name} {last_name}!"


def power(base, exponent=2):
  return base**exponent


def multiply(*args):
  result = 1
  for num in args:
    result *= num
  return result


def describe_person(**kwargs):
  description = ", ".join(f"{key}: {value}" for key, value in kwargs.items())
  return f"Person details: {description}"


sum_result = add(5, 10)

greeting = greet(first_name="John", last_name="Doe")

square = power(5)
cube = power(5, 3)

product = multiply(2, 3, 4)

person_description = describe_person(name="Alice", age=30, city="New York")

print("Sum:", sum_result)
print("Greeting:", greeting)
print("Square:", square)
print("Cube:", cube)
print("Product:", product)
print(person_description)
```

```python

def add(a, b):
  return a + b


def subtract(a, b):
  return a - b

def area_of_circle(radius):
  return 3.14 * radius * radius

def perimeter_of_circle(radius):
  return 2 * 3.14 * radius

sum_result = add(10, 5)
diff_result = subtract(10, 5)

area = area_of_circle(7)
perimeter = perimeter_of_circle(7)

print("Sum:", sum_result)
print("Difference:", diff_result)
print("Area of Circle:", area)
print("Perimeter of Circle:", perimeter)
```

```python
import shutil

source_file = "source.txt"
destination_file = "destination.txt"
append_file = "append_file.txt"

try:
  with open(source_file, "w") as file:
    file.write("This is the original content.\n")
    file.write("Adding some more text.")

  with open(source_file, "r") as file:
    content = file.read()
    print("Read content from source file:")
    print(content)
except IOError as e:
  print(f"Error opening/reading file: {e}")

try:

  with open(source_file, "a") as file:
    file.write("\nAppending new content.")

  with open(source_file, "r") as file:
    content = file.read()
    print("\nContent after appending:")
    print(content)
except IOError as e:
  print(f"Error opening/reading/appending file: {e}")

try:
 
  shutil.copy(source_file, destination_file)
  print(f"\nFile copied from {source_file} to {destination_file}.")
    
  with open(destination_file, "r") as file:
    content = file.read()
    print("\nContent of the destination file:")
    print(content)
except IOError as e:
  print(f"Error copying file: {e}")
```

```python

def divide_numbers(num1, num2):
  try:
   
    result = num1 / num2
    print(f"The result is: {result}")
  except ZeroDivisionError:
   
    print("Error: Cannot divide by zero!")
  except TypeError:
    
    print("Error: Invalid input type. Please enter numbers only.")
  except ValueError:
    
    print("Error: Invalid input value. Please enter a valid number.")
  except Exception as e:
   
    print(f"An unexpected error occurred: {e}")

print("Test 1: Valid inputs")
divide_numbers(10, 2)

print("\nTest 2: Division by zero")
divide_numbers(10, 0)

print("\nTest 3: Invalid input types")
divide_numbers("ten", 2)

print("\nTest 4: Value error")
divide_numbers(10, "two")

print("\nTest 5: Unexpected exception")
divide_numbers(10, None)


```
